## 1. 什么是融合范畴？

### 定义

**融合范畴**是一个数学结构，包含：

1. **对象集合** $\{a, b, c, ...\}$
2. **融合规则** $a \times b = \sum_c N_{ab}^c \, c$
3. **单位对象** $1$，满足 $1 \times a = a \times 1 = a$
4. **结合性** $(a \times b) \times c \cong a \times (b \times c)$

### 物理意义

- **对象**：任意子类型、对称性算符
- **融合**：粒子碰撞、算符组合
- **量子维数**：自由度数量

## 2. 阿贝尔范畴：Z₂

最简单的非平凡例子。

```python
# 创建Z₂范畴
z2 = Z2Category()
z2.print_info()

# 可视化融合规则
def visualize_z2_fusion():
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # 创建融合表
    objects = ['1', 'e']
    fusion_table = [
        ['1', 'e'],
        ['e', '1']
    ]
    
    # 绘制表格
    table_data = [['×'] + objects]
    for i, row in enumerate(fusion_table):
        table_data.append([objects[i]] + row)
    
    table = ax.table(cellText=table_data, cellLoc='center',
                    bbox=[0.2, 0.5, 0.6, 0.4])
    table.auto_set_font_size(False)
    table.set_fontsize(14)
    table.scale(1, 2)
    
    # 颜色
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0 or j == 0:
                cell.set_facecolor('lightgray')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('lightblue' if table_data[i][j] == '1' else 'lightcoral')
    
    ax.set_title('Z₂ Fusion Rules (Abelian)\n群乘法表', 
                fontsize=16, fontweight='bold')
    ax.axis('off')
    
    # 添加说明
    ax.text(0.5, 0.3, 'Properties:', ha='center', fontsize=13, fontweight='bold',
           transform=ax.transAxes)
    ax.text(0.5, 0.22, '• Abelian: a×b = b×a', ha='center', fontsize=11,
           transform=ax.transAxes)
    ax.text(0.5, 0.16, '• Invertible: e×e = 1', ha='center', fontsize=11,
           transform=ax.transAxes)
    ax.text(0.5, 0.10, '• All d_a = 1 (trivial dimensions)', ha='center', fontsize=11,
           transform=ax.transAxes)
    
    plt.tight_layout()
    return fig

visualize_z2_fusion()
plt.show()
```

## 3. 非阿贝尔范畴：Fibonacci

第一个真正有趣的例子！

```python
# 创建Fibonacci范畴
fib = FibonacciCategory()
fib.print_info()

# 黄金比例的重要性
print("\n黄金比例 φ 的特殊性质:")
print("="*50)
phi = fib.phi
print(f"φ = {phi:.10f}")
print(f"φ² = {phi**2:.10f}")
print(f"φ + 1 = {phi + 1:.10f}")
print(f"φ² - φ - 1 = {phi**2 - phi - 1:.10e}  (应为0!)")
print(f"\n这意味着: φ² = φ + 1")
print(f"对应融合规则: τ × τ = 1 + τ")
```

```python
# 可视化Fibonacci融合
def visualize_fibonacci_fusion():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    
    # 左图：融合树
    ax1.set_xlim(0, 10)
    ax1.set_ylim(0, 8)
    ax1.axis('off')
    ax1.set_title('Fibonacci Fusion Tree: τ×τ×τ', 
                 fontsize=14, fontweight='bold')
    
    # 第一步: τ × τ = 1 + τ
    ax1.text(2, 7, 'Step 1: τ × τ = 1 + τ', fontsize=12, fontweight='bold')
    
    # 分支1: τ × τ → 1
    circle1 = Circle((2, 5.5), 0.3, color='orange', ec='black', linewidth=2)
    ax1.add_patch(circle1)
    ax1.text(2, 5.5, 'τ', ha='center', va='center', fontsize=11, fontweight='bold')
    
    circle2 = Circle((3.5, 5.5), 0.3, color='orange', ec='black', linewidth=2)
    ax1.add_patch(circle2)
    ax1.text(3.5, 5.5, 'τ', ha='center', va='center', fontsize=11, fontweight='bold')
    
    ax1.plot([2.3, 2.7], [5.3, 5], 'k-', linewidth=2)
    ax1.plot([3.2, 2.8], [5.3, 5], 'k-', linewidth=2)
    
    circle3 = Circle((2.75, 4.7), 0.3, color='lightblue', ec='black', linewidth=2)
    ax1.add_patch(circle3)
    ax1.text(2.75, 4.7, '1+τ', ha='center', va='center', fontsize=10, fontweight='bold')
    
    # 第二步
    ax1.text(2, 3.5, 'Step 2: (1+τ) × τ', fontsize=12, fontweight='bold')
    
    # 分支1: 1 × τ = τ
    ax1.text(1, 2.5, '1×τ=τ', fontsize=10, bbox=dict(boxstyle='round', 
            facecolor='lightyellow'))
    
    # 分支2: τ × τ = 1 + τ
    ax1.text(4, 2.5, 'τ×τ=1+τ', fontsize=10, bbox=dict(boxstyle='round',
            facecolor='lightgreen'))
    
    # 最终结果
    ax1.text(2.5, 1.2, 'τ×τ×τ = τ + (1+τ) = 1 + 2τ', ha='center',
            fontsize=13, fontweight='bold',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    # 右图：量子维数
    objects = ['1', 'τ']
    dims = [1, phi]
    
    bars = ax2.bar(objects, dims, color=['lightblue', 'orange'],
                  edgecolor='black', linewidth=2, alpha=0.7)
    
    # 标注数值
    for bar, val in zip(bars, dims):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height,
                f'{val:.4f}',
                ha='center', va='bottom', fontsize=12, fontweight='bold')
    
    # 总量子维数
    D = np.sqrt(1 + phi**2)
    ax2.axhline(D, color='red', linestyle='--', linewidth=2,
               label=f'𝒟 = √(1² + φ²) = {D:.4f}')
    
    ax2.set_ylabel('Quantum Dimension', fontsize=12)
    ax2.set_title('Quantum Dimensions\n(Non-trivial!)', 
                 fontsize=14, fontweight='bold')
    ax2.legend(fontsize=11)
    ax2.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    return fig

visualize_fibonacci_fusion()
plt.show()
```

## 4. F-符号：结合性的代数

### 为什么需要F-符号？

融合是结合的，但不同的结合顺序对应不同的基：

$$
(a \times b) \times c \quad \text{vs} \quad a \times (b \times c)
$$

F-符号描述这两种基之间的变换。

### Fibonacci F-符号

```python
# Fibonacci F-符号
F_matrix = fib.F_tau_tau_tau

print("Fibonacci F-符号: F^{τττ}_{τττ}")
print("="*50)
print(F_matrix)
print(f"\n矩阵元素:")
print(f"F[0,0] = φ^(-1) = {F_matrix[0,0]:.6f}")
print(f"F[0,1] = φ^(-1/2) = {F_matrix[0,1]:.6f}")
print(f"F[1,0] = φ^(-1/2) = {F_matrix[1,0]:.6f}")
print(f"F[1,1] = -φ^(-1) = {F_matrix[1,1]:.6f}")

# 验证酉性
print(f"\n验证酉性: F† F = I")
result = F_matrix.conj().T @ F_matrix
print(result)
error = np.linalg.norm(result - np.eye(2))
print(f"\n误差: {error:.2e}")
if error < 1e-10:
    print("✓ F-符号是酉的！")

# 可视化
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# F-符号矩阵
im1 = ax1.imshow(np.abs(F_matrix), cmap='RdBu_r', vmin=0, vmax=1)
ax1.set_xticks([0, 1])
ax1.set_yticks([0, 1])
ax1.set_xticklabels(['|1⟩', '|τ⟩'])
ax1.set_yticklabels(['⟨1|', '⟨τ|'])
ax1.set_title('|F^{τττ}_{τττ}|', fontsize=14, fontweight='bold')

for i in range(2):
    for j in range(2):
        text = ax1.text(j, i, f'{F_matrix[i,j]:.4f}',
                       ha="center", va="center", color="black", fontsize=11)

plt.colorbar(im1, ax=ax1)

# 酉性验证
im2 = ax2.imshow(np.abs(result), cmap='Greens', vmin=0, vmax=1)
ax2.set_title('|F† F| (Should be Identity)', fontsize=14, fontweight='bold')

for i in range(2):
    for j in range(2):
        text = ax2.text(j, i, f'{result[i,j]:.4f}',
                       ha="center", va="center", color="black", fontsize=11)

plt.colorbar(im2, ax=ax2)

plt.tight_layout()
plt.show()
```

## 5. 五边形方程 (Pentagon Equation)

### 一致性条件

F-符号必须满足**五边形方程**，这保证了结合性的一致性：

$$
\sum_n F_{abe}^{cdn} F_{cdf}^{emn} = \sum_p F_{bcd}^{fep} F_{apc}^{emf}
$$

这对应于4个对象的5种不同结合方式。

```python
def visualize_pentagon():
    """可视化五边形方程"""
    fig, ax = plt.subplots(figsize=(10, 10))
    
    # 五边形的顶点（5种结合方式）
    angles = np.linspace(np.pi/2, np.pi/2 + 2*np.pi, 6)
    radius = 3
    
    vertices = [(radius * np.cos(a), radius * np.sin(a)) for a in angles[:-1]]
    
    # 标签
    labels = [
        '((a×b)×c)×d',
        '(a×b)×(c×d)',
        'a×(b×(c×d))',
        'a×((b×c)×d)',
        '(a×(b×c))×d'
    ]
    
    # 绘制五边形
    for i in range(5):
        j = (i + 1) % 5
        ax.plot([vertices[i][0], vertices[j][0]], 
               [vertices[i][1], vertices[j][1]], 
               'k-', linewidth=2)
    
    # 绘制顶点和标签
    for i, (v, label) in enumerate(zip(vertices, labels)):
        circle = Circle(v, 0.2, color='lightblue', ec='black', linewidth=2, zorder=3)
        ax.add_patch(circle)
        
        # 标签位置
        offset = 1.3
        label_pos = (v[0] * offset, v[1] * offset)
        ax.text(label_pos[0], label_pos[1], label, 
               ha='center', va='center', fontsize=11,
               bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    # 标注F-移动
    ax.text(0, 0, 'Pentagon\nEquation', ha='center', va='center',
           fontsize=14, fontweight='bold',
           bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))
    
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title('Pentagon Equation\n5 ways to associate 4 objects',
                fontsize=14, fontweight='bold', pad=20)
    
    plt.tight_layout()
    return fig

visualize_pentagon()
plt.show()

print("\n五边形方程的物理意义:")
print("="*60)
print("• 保证融合的一致性")
print("• 任何4个粒子，无论如何结合，最终结果相同")
print("• 是融合范畴的核心公理")
print("• 对Fibonacci: 可以验证F-符号确实满足此方程！")
```

## 6. 比较不同融合范畴

```python
# 比较表
categories_info = {
    'Z₂': {
        'objects': 2,
        'D': 2.0,
        'abelian': True,
        'all_d=1': True,
        'application': 'Toric Code'
    },
    'Fibonacci': {
        'objects': 2,
        'D': phi * np.sqrt(2),
        'abelian': False,
        'all_d=1': False,
        'application': 'Universal TQC'
    },
    'Ising': {
        'objects': 3,
        'D': 2.0,
        'abelian': False,
        'all_d=1': False,
        'application': 'Majorana fermions'
    },
    'SU(2)₂': {
        'objects': 3,
        'D': 2.0,
        'abelian': False,
        'all_d=1': False,
        'application': 'Quantum groups'
    }
}

print("\n融合范畴比较")
print("="*80)
print(f"{'Category':<15} {'Objects':<10} {'𝒟':<12} {'Abelian':<10} {'All d=1':<10} {'Application':<20}")
print("-"*80)

for name, info in categories_info.items():
    print(f"{name:<15} {info['objects']:<10} {info['D']:<12.4f} "
          f"{'Yes' if info['abelian'] else 'No':<10} "
          f"{'Yes' if info['all_d=1'] else 'No':<10} "
          f"{info['application']:<20}")

# 可视化
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# 总量子维数
names = list(categories_info.keys())
D_values = [info['D'] for info in categories_info.values()]
colors = ['blue', 'orange', 'green', 'purple']

bars1 = ax1.bar(names, D_values, color=colors, edgecolor='black', 
               linewidth=2, alpha=0.7)
ax1.set_ylabel('Total Quantum Dimension 𝒟', fontsize=12)
ax1.set_title('Total Quantum Dimensions', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='y')
ax1.axhline(2, color='red', linestyle='--', linewidth=1.5, alpha=0.5)

# 标注
for bar, val in zip(bars1, D_values):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
            f'{val:.3f}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# 拓扑纠缠熵
S_topo = [np.log(D) for D in D_values]

bars2 = ax2.bar(names, S_topo, color=colors, edgecolor='black',
               linewidth=2, alpha=0.7)
ax2.set_ylabel('Topological Entropy ln(𝒟)', fontsize=12)
ax2.set_title('Topological Entanglement Entropy', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars2, S_topo):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{val:.3f}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```
