## 1. String-Net态构造

### 基本思想

String-Net基态是格子上所有满足融合规则的弦网配置的叠加：

$$
|\psi\rangle = \sum_{\text{configs}} w(\text{config}) |\text{config}\rangle
$$

权重 $w$ 由量子维数和F-符号决定。

```python
def visualize_string_net():
    """可视化String-Net配置"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
    
    # 左图：单个配置
    ax1.set_xlim(0, 8)
    ax1.set_ylim(0, 8)
    ax1.set_aspect('equal')
    ax1.axis('off')
    ax1.set_title('Single String-Net Configuration', 
                 fontsize=14, fontweight='bold')
    
    # 蜂窝格（简化）
    vertices = [
        (2, 4), (3, 5), (4, 4), (3, 3),  # 中心六边形
        (3, 6), (5, 5), (5, 3), (3, 2), (1, 3), (1, 5)  # 外围
    ]
    
    edges_with_labels = [
        ((2, 4), (3, 5), 'τ'),
        ((3, 5), (4, 4), '1'),
        ((4, 4), (3, 3), 'τ'),
        ((3, 3), (2, 4), '1'),
        ((3, 5), (3, 6), 'τ'),
        ((4, 4), (5, 5), 'τ'),
        ((3, 3), (3, 2), '1'),
        ((2, 4), (1, 3), 'τ'),
    ]
    
    # 绘制边（弦）
    for v1, v2, label in edges_with_labels:
        color = 'blue' if label == '1' else 'red'
        linewidth = 2 if label == '1' else 3
        ax1.plot([v1[0], v2[0]], [v1[1], v2[1]], 
                color=color, linewidth=linewidth, zorder=1)
        
        # 标签
        mid_x = (v1[0] + v2[0]) / 2
        mid_y = (v1[1] + v2[1]) / 2
        ax1.text(mid_x, mid_y, label, fontsize=11, fontweight='bold',
                bbox=dict(boxstyle='circle', facecolor='yellow' if label=='1' else 'orange',
                         alpha=0.8), zorder=3)
    
    # 绘制顶点
    for v in vertices[:4]:  # 中心顶点
        circle = Circle(v, 0.15, color='lightblue', ec='black', 
                       linewidth=2, zorder=2)
        ax1.add_patch(circle)
    
    ax1.text(4, 7, 'String labels: 1, τ', fontsize=11,
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.7))
    
    # 右图：叠加态
    ax2.set_xlim(0, 10)
    ax2.set_ylim(0, 8)
    ax2.axis('off')
    ax2.set_title('String-Net Superposition State', 
                 fontsize=14, fontweight='bold')
    
    # 表示叠加
    configs = ['Config 1', 'Config 2', 'Config 3', '...']
    weights = ['w₁', 'w₂', 'w₃', '']
    
    y_start = 6
    for i, (config, weight) in enumerate(zip(configs, weights)):
        y = y_start - i * 1.5
        
        if weight:
            ax2.text(1, y, weight, fontsize=12, fontweight='bold')
            ax2.text(2, y, '×', fontsize=12)
        
        rect = Rectangle((2.5, y-0.3), 3, 0.6, 
                        facecolor='lightblue' if i % 2 == 0 else 'lightcoral',
                        edgecolor='black', linewidth=2)
        ax2.add_patch(rect)
        ax2.text(4, y, config, ha='center', va='center', fontsize=11)
        
        if i < len(configs) - 1:
            ax2.text(6, y, '+', fontsize=14, fontweight='bold')
    
    # 说明
    ax2.text(5, 1.5, 'Weights determined by:', ha='center',
            fontsize=11, fontweight='bold')
    ax2.text(5, 1.0, '• Quantum dimensions d_a', ha='center', fontsize=10)
    ax2.text(5, 0.5, '• F-symbols', ha='center', fontsize=10)
    
    plt.tight_layout()
    return fig

visualize_string_net()
plt.show()
```

## 2. 任意子激发

### 创建任意子

在String-Net基态上，激发对应融合范畴的对象：

- **真空**: 1
- **τ任意子**: τ (对于Fibonacci)

### 任意子的性质

```python
# Fibonacci任意子
fib = FibonacciCategory()

print("Fibonacci任意子")
print("="*60)
print("\n任意子类型: 1, τ")
print(f"\n量子维数:")
print(f"  d₁ = {fib.quantum_dims['1']}")
print(f"  d_τ = {fib.quantum_dims['tau']:.6f} = φ (黄金比例!)")

print(f"\n融合通道:")
for (a, b), results in fib.fusion_rules.items():
    result_str = ' + '.join(results)
    print(f"  {a} × {b} = {result_str}")

print(f"\nτ × τ的融合:")
print(f"  结果: 1 + τ (两个通道!)")
print(f"  这意味着τ是非阿贝尔任意子")

# 可视化
def visualize_anyon_fusion():
    fig, ax = plt.subplots(figsize=(12, 8))
    
    ax.set_xlim(0, 12)
    ax.set_ylim(0, 10)
    ax.axis('off')
    ax.set_title('Fibonacci Anyon Fusion: τ × τ = 1 + τ',
                fontsize=16, fontweight='bold')
    
    # 初始态：两个τ任意子
    ax.text(6, 9, 'Initial State', ha='center', fontsize=13, fontweight='bold')
    
    tau1 = Circle((4, 7.5), 0.4, color='orange', ec='black', linewidth=3)
    tau2 = Circle((8, 7.5), 0.4, color='orange', ec='black', linewidth=3)
    ax.add_patch(tau1)
    ax.add_patch(tau2)
    ax.text(4, 7.5, 'τ', ha='center', va='center', 
           fontsize=14, fontweight='bold', color='white')
    ax.text(8, 7.5, 'τ', ha='center', va='center',
           fontsize=14, fontweight='bold', color='white')
    
    # 箭头
    arrow1 = FancyArrowPatch((3, 6.5), (2, 4.5), 
                            arrowstyle='->', lw=3, color='blue')
    arrow2 = FancyArrowPatch((9, 6.5), (10, 4.5),
                            arrowstyle='->', lw=3, color='red')
    ax.add_patch(arrow1)
    ax.add_patch(arrow2)
    
    # 融合结果
    ax.text(6, 5.5, 'Fusion Outcomes', ha='center', 
           fontsize=13, fontweight='bold')
    
    # 通道1: → 1
    ax.text(2, 3.5, 'Channel 1', ha='center', fontsize=11, fontweight='bold')
    vacuum = Circle((2, 2.5), 0.4, color='lightblue', ec='black', linewidth=3)
    ax.add_patch(vacuum)
    ax.text(2, 2.5, '1', ha='center', va='center',
           fontsize=14, fontweight='bold')
    ax.text(2, 1.5, 'Vacuum\n(trivial)', ha='center', fontsize=9)
    
    # 通道2: → τ
    ax.text(10, 3.5, 'Channel 2', ha='center', fontsize=11, fontweight='bold')
    tau_result = Circle((10, 2.5), 0.4, color='orange', ec='black', linewidth=3)
    ax.add_patch(tau_result)
    ax.text(10, 2.5, 'τ', ha='center', va='center',
           fontsize=14, fontweight='bold', color='white')
    ax.text(10, 1.5, 'Another τ\n(non-trivial)', ha='center', fontsize=9)
    
    # 说明框
    ax.text(6, 0.5, '⚠ Non-Abelian: Multiple fusion channels!',
           ha='center', fontsize=12, fontweight='bold',
           bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.8))
    
    plt.tight_layout()
    return fig

visualize_anyon_fusion()
plt.show()
```

## 3. 编织操作

### 什么是编织？

在2D空间中，交换两个任意子的轨迹不是平凡的！

$$
R_{ab}: |a, b\rangle \to e^{i\theta_{ab}} |b, a\rangle
$$

### 编织相位

```python
def visualize_braiding():
    """可视化任意子编织"""
    fig, axes = plt.subplots(2, 3, figsize=(16, 10))
    
    # 时间演化序列
    times = ['t=0', 't=1', 't=2', 't=3', 't=4', 't=5']
    
    for idx, (ax, time) in enumerate(zip(axes.flat, times)):
        ax.set_xlim(0, 10)
        ax.set_ylim(0, 8)
        ax.set_aspect('equal')
        ax.axis('off')
        ax.set_title(time, fontsize=12, fontweight='bold')
        
        if idx == 0:
            # 初始位置
            tau1 = Circle((3, 4), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((7, 4), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(3, 4, 'τ₁', ha='center', va='center', 
                   fontsize=11, fontweight='bold', color='white')
            ax.text(7, 4, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            
        elif idx == 1:
            # τ₁向右移动
            tau1 = Circle((5, 4), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((7, 4), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(5, 4, 'τ₁', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            ax.text(7, 4, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            # 轨迹
            ax.plot([3, 5], [4, 4], 'orange', linewidth=2, 
                   linestyle='--', alpha=0.5)
            
        elif idx == 2:
            # τ₁在上方，开始交叉
            tau1 = Circle((6, 5.5), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((7, 4), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(6, 5.5, 'τ₁', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            ax.text(7, 4, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            # 轨迹
            t = np.linspace(0, 1, 20)
            x = 5 + t
            y = 4 + 1.5*t
            ax.plot(x, y, 'orange', linewidth=2, linestyle='--', alpha=0.5)
            
        elif idx == 3:
            # 完成交叉
            tau1 = Circle((7, 5.5), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((7, 4), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(7, 5.5, 'τ₁', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            ax.text(7, 4, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            
        elif idx == 4:
            # τ₁向右移动
            tau1 = Circle((8, 4.5), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((6, 3.5), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(8, 4.5, 'τ₁', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            ax.text(6, 3.5, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            
        else:
            # 最终位置（已交换）
            tau1 = Circle((7, 4), 0.4, color='orange', ec='black', lw=2)
            tau2 = Circle((3, 4), 0.4, color='red', ec='black', lw=2)
            ax.add_patch(tau1)
            ax.add_patch(tau2)
            ax.text(7, 4, 'τ₁', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            ax.text(3, 4, 'τ₂', ha='center', va='center',
                   fontsize=11, fontweight='bold', color='white')
            
            # 标注相位
            ax.text(5, 6, 'Result: e^(iθ)|τ₂, τ₁⟩', ha='center',
                   fontsize=11, fontweight='bold',
                   bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.8))
    
    plt.suptitle('Braiding Two Fibonacci Anyons', 
                fontsize=16, fontweight='bold', y=0.98)
    plt.tight_layout()
    return fig

visualize_braiding()
plt.show()

print("\n编织统计的关键性质:")
print("="*60)
print("• 非交换: R₁₂ R₂₁ ≠ I (一般)")
print("• 非阿贝尔任意子: 编织依赖于融合通道")
print("• 拓扑保护: 只依赖于编织拓扑，不依赖于具体路径")
print("• 量子计算: 编织操作 = 量子门")
```

## 4. R-矩阵和编织相位

### Fibonacci R-矩阵

对于 $\tau \times \tau$ 系统（两个融合通道：1和τ）：

$$
R_\tau^\tau = \begin{pmatrix}
e^{4\pi i/5} & 0 \\
0 & e^{-3\pi i/5}
\end{pmatrix}
$$

```python
# Fibonacci R-矩阵
theta_1 = 4 * np.pi / 5
theta_tau = -3 * np.pi / 5

R_matrix = np.array([
    [np.exp(1j * theta_1), 0],
    [0, np.exp(1j * theta_tau)]
])

print("Fibonacci R-矩阵")
print("="*60)
print(f"\nR_τ^τ (编织两个τ任意子):")
print(R_matrix)

print(f"\n编织相位:")
print(f"  通道 1: θ₁ = 4π/5 = {theta_1:.6f} rad = {np.degrees(theta_1):.2f}°")
print(f"  通道 τ: θ_τ = -3π/5 = {theta_tau:.6f} rad = {np.degrees(theta_tau):.2f}°")

print(f"\n相位差:")
print(f"  Δθ = θ₁ - θ_τ = 7π/5 = {(theta_1 - theta_tau):.6f} rad")
print(f"  这是非阿贝尔统计的标志！")

# 可视化R-矩阵
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 5))

# R-矩阵幅值
im1 = ax1.imshow(np.abs(R_matrix), cmap='Blues', vmin=0, vmax=1)
ax1.set_xticks([0, 1])
ax1.set_yticks([0, 1])
ax1.set_xticklabels(['|1⟩', '|τ⟩'])
ax1.set_yticklabels(['⟨1|', '⟨τ|'])
ax1.set_title('|R| (Magnitude)', fontsize=13, fontweight='bold')
for i in range(2):
    for j in range(2):
        ax1.text(j, i, f'{np.abs(R_matrix[i,j]):.2f}',
                ha="center", va="center", fontsize=11)
plt.colorbar(im1, ax=ax1)

# 相位
phases = np.angle(R_matrix)
im2 = ax2.imshow(phases, cmap='twilight', vmin=-np.pi, vmax=np.pi)
ax2.set_xticks([0, 1])
ax2.set_yticks([0, 1])
ax2.set_xticklabels(['|1⟩', '|τ⟩'])
ax2.set_yticklabels(['⟨1|', '⟨τ|'])
ax2.set_title('arg(R) (Phase)', fontsize=13, fontweight='bold')
for i in range(2):
    for j in range(2):
        if np.abs(R_matrix[i,j]) > 0.01:
            ax2.text(j, i, f'{phases[i,j]:.3f}',
                    ha="center", va="center", fontsize=10)
plt.colorbar(im2, ax=ax2, label='Phase (rad)')

# 复平面
ax3.set_xlim(-1.5, 1.5)
ax3.set_ylim(-1.5, 1.5)
ax3.set_aspect('equal')
ax3.axhline(0, color='k', linewidth=0.5)
ax3.axvline(0, color='k', linewidth=0.5)
ax3.set_xlabel('Re', fontsize=11)
ax3.set_ylabel('Im', fontsize=11)
ax3.set_title('Phases in Complex Plane', fontsize=13, fontweight='bold')

# 单位圆
theta_circle = np.linspace(0, 2*np.pi, 100)
ax3.plot(np.cos(theta_circle), np.sin(theta_circle), 
        'k--', linewidth=1, alpha=0.3)

# 相位向量
for i, (theta, label, color) in enumerate([(theta_1, '|1⟩', 'blue'),
                                            (theta_tau, '|τ⟩', 'red')]):
    x, y = np.cos(theta), np.sin(theta)
    ax3.arrow(0, 0, x*0.9, y*0.9, head_width=0.1, head_length=0.1,
             fc=color, ec=color, linewidth=2)
    ax3.text(x*1.2, y*1.2, label, ha='center', va='center',
            fontsize=11, fontweight='bold',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. 拓扑量子计算

### 编织作为量子门

Fibonacci任意子的编织可以实现**通用量子计算**！

关键思想：
1. 编码：量子信息存储在anyons的融合通道
2. 门操作：通过编织实现
3. 读出：测量最终融合结果

```python
def demonstrate_tqc():
    """演示拓扑量子计算"""
    print("拓扑量子计算示例")
    print("="*60)
    
    print("\n1. 量子比特编码")
    print("   使用3个τ任意子:")
    print("   τ₁ × τ₂ × τ₃")
    print("   ")
    print("   融合结果: (τ×τ)×τ = (1+τ)×τ = 1+2τ")
    print("   两个τ通道 → 2维希尔伯特空间 → 1个量子比特!")
    
    print("\n2. 量子门")
    print("   编织 τ₁ 和 τ₂:")
    print("   |ψ⟩ → R₁₂ |ψ⟩")
    print("   ")
    print("   不同编织序列 → 不同酉操作")
    print("   可以近似任意单比特门！")
    
    print("\n3. 拓扑保护")
    print("   • 信息存储在全局拓扑性质")
    print("   • 局域噪声无法破坏")
    print("   • 天然容错")
    
    print("\n4. 通用性证明")
    print("   • Fibonacci anyons是通用的 (Freedman et al., 2002)")
    print("   • 可以近似SU(2)的稠密子集")
    print("   • Solovay-Kitaev定理保证高效编译")
    
    # 可视化
    fig, ax = plt.subplots(figsize=(12, 8))
    
    ax.set_xlim(0, 12)
    ax.set_ylim(0, 10)
    ax.axis('off')
    ax.set_title('Topological Quantum Computation with Fibonacci Anyons',
                fontsize=14, fontweight='bold')
    
    # 3个anyons
    positions = [(2, 6), (6, 6), (10, 6)]
    for i, pos in enumerate(positions, 1):
        circle = Circle(pos, 0.5, color='orange', ec='black', lw=2)
        ax.add_patch(circle)
        ax.text(pos[0], pos[1], f'τ{i}', ha='center', va='center',
               fontsize=13, fontweight='bold', color='white')
    
    # 融合通道
    ax.text(4, 5, '×', fontsize=16, fontweight='bold')
    ax.text(8, 5, '×', fontsize=16, fontweight='bold')
    
    # 结果
    ax.text(6, 3.5, 'Fusion Channels', ha='center',
           fontsize=12, fontweight='bold')
    
    ax.text(4, 2.5, '|0⟩: ...→1→τ', ha='center',
           fontsize=11,
           bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))
    ax.text(8, 2.5, '|1⟩: ...→τ→τ', ha='center',
           fontsize=11,
           bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))
    
    # 编织门
    ax.text(6, 1, 'Braiding = Quantum Gate', ha='center',
           fontsize=13, fontweight='bold',
           bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

demonstrate_tqc()
```
